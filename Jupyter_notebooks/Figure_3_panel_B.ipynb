# convert up function # from https://karobben.github.io/2021/03/31/R/ggupset/ and https://github.com/const-ae/ggupset

```r
Convert_up <- function(TB){
  
  
  TB_tmp = TB
  
  cat("TB_tmp_initial\n")
  cat(str(TB_tmp))
  cat("\n")
  
  for(i in colnames(TB)){
    # cat("-------------------------->i\t")
    # cat(sprintf(as.character(i)))
    # cat("\n")
    
    TB_tmp[i] = i
    
    # cat("TB_tmp_0\n")
    # cat(str(TB_tmp))
    # cat("\n")
    
  }
  TB_tmp[TB == 0] = ""
  
  # cat("TB_tmp_1\n")
  # cat(str(TB_tmp))
  # cat("\n")
  
  TB_t <- data.frame(t(TB_tmp), stringsAsFactors = F)
  
  # cat("TB_t_0\n")
  # cat(str(TB_t))
  # cat("\n")
  
  TB_tmp$upset = as.list(TB_t)
  
  # cat("TB_tmp_FIN\n")
  # cat(str(TB_tmp))
  # cat("\n")
  
  return(TB_tmp)
}
```
